복습: (조건1) & (조건2) 괄호 필수!

핵심: groupby('기준')['대상'].mean() -> "기준끼리 묶어서 계산해줘"

````python
import pandas as pd
````

````python
data = {
      '반' : ['A반', 'A반','B반', 'B반', 'C반', 'C반' ]
    ,'이름' : ['철수', '영희', '민수', '지민', '구름', '바람']
    ,'국어' : [90, 80, 70, 60, 100, 90]
    ,'수학': [85, 95, 80, 70, 90, 80]
}

df_score = pd.DataFrame(data)
print("학생 성적 데이터 준비 완료")
print(df_score)

df_A_korean_85 = df_score[(df_score['반'] == 'A반') & (df_score['국어']>=85)]

print('-'*30)

print(df_A_korean_85)
````

````python
# 연습문제 1
avg_score_korea = df_score.groupby('반')['국어'].mean() #""반별""로 구하는 것임
print(avg_score_korea)

# 연습문제 2
count_student = df_score.groupby('반')['이름']
print(count_student.count())
````
