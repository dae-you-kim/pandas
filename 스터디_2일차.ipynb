```python
import pandas as pd
```

```python
data_exam = {
    '이름' : ['철수', '지민', '현수', '민지'],
    '국어' : [80,90,100,85],
    '수학' : [100,85,75,85]
}
```

```python
df_exam = pd.DataFrame(data_exam)

print(df_exam)
```

```python
kor_mean = df_exam['국어'].mean()
print("국어 평균점수 =",kor_mean)
```

```python
data_sales = {
    '직원': ['김대리', '이과장', '박사원', '최부장', '정인턴'],
    '부서': ['영업1팀', '영업2팀', '영업1팀', '영업2팀', '영업1팀'],
    '판매액': [500, 1200, 300, 2000, 100],  # 단위: 만원
    '근무년수': [3, 7, 1, 15, 0]
}

df_sales = pd.DataFrame(data_sales)
print(df_sales)
print("영업팀 실적 준비 완료")
```

```python
#1교시. 필터링하기
df_veteran = df_sales[df_sales['근무년수']>=5]
print(df_veteran)
```

```python
#day2_미션 1 : 이번엔 **'판매액'이 1000만 원 이상(>=)**인 고성과자 직원들만 뽑아서 확인해 보세요.
df_highsalaries = df_sales[df_sales['판매액']>=1000]

print(df_highsalaries)
```

```python
#2교시. "순서"대로 줄 세우기 (Sorting)
#[👨‍🏫 선생님의 시범] 근무년수가 긴 순서대로(내림차순) 정렬해볼까요?
df_sales.sort_values(by='근무년수',ascending=False)
```

```python
#✍️ [Day 2 - 미션 2] 문제: 이번 달 '판매액'이 가장 높은 순서대로(1등이 맨 위에 오게) 직원들을 줄 세워 보세요.
df_sales.sort_values(by='판매액',ascending=False)

df_sorted = df_sales.sort_values(by='판매액',ascending=False)

print(df_sorted)

```

```python
#3교시. 까다로운 조건 (AND 연산자)
#[👨‍🏫 선생님의 시범] '영업2팀'이면서 '근무년수가 10년 이상'인 사람을 찾아봅시다.

df_and = df_sales[(df_sales['부서'] == '영업2팀') & (df_sales['근무년수']>=10)]
print(df_and)
```

```python
#문제: 조건 1: 부서가 '영업1팀' 이고 (== '영업1팀')//조건 2: 판매액이 400만 원 이상인 (>= 400) 직원을 찾아주세요.

df_winner = df_sales[(df_sales['부서'] == '영업1팀') & (df_sales['판매액'] >= 400)]
print(df_winner)
```
